```python
class Product:
    def __init__(self, price):
        self._price = price

    # Getter method
    def get_price(self):
        return self._price

    # Setter method
    def set_price(self, new_price):
        if new_price < 0:
            raise ValueError("Price cannot be negative")
        self._price = new_price


# Creating an object
product = Product(1000)

print("Original Price:", product.get_price())

# Updating price
product.set_price(1200)
print("Updated Price:", product.get_price())

# Trying to set negative price
# product.set_price(-500)
```

```python
class Playlist:
    def __init__(self):
        self._songs = []

    # Add a song
    def add_song(self, song):
        self._songs.append(song)

    # Remove a song
    def remove_song(self, song):
        if song in self._songs:
            self._songs.remove(song)
        else:
            print("Song not found")

    # Getter method
    def get_songs(self):
        return self._songs


# Creating a Playlist object
playlist = Playlist()

# Adding songs
playlist.add_song("Tum Hi Ho")
playlist.add_song("Kesariya")
playlist.add_song("Chaleya")

print("Songs:", playlist.get_songs())

# Removing a song
playlist.remove_song("Kesariya")

print("Updated Songs:", playlist.get_songs())
```

```python
from abc import ABC, abstractmethod


# Abstract class
class PaymentMethod(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


# UPI subclass
class UPI(PaymentMethod):

    def pay(self, amount):
        print(f"Processing UPI payment of ₹{amount}")


# CreditCard subclass
class CreditCard(PaymentMethod):

    def pay(self, amount):
        print(f"Processing Credit Card payment of ₹{amount}")


# Creating objects
upi_payment = UPI()
card_payment = CreditCard()

# Making payments
upi_payment.pay(500)
card_payment.pay(1500)
```
